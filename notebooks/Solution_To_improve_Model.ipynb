```python
import torch
import torch.nn as nn
from torchsummary import summary
```

# Batch Normalization

```python
model = nn.Sequential(
    nn.Conv2d(in_channels=1, out_channels=3, kernel_size=3, padding='same', stride=1),
    nn.BatchNorm2d(num_features=3),
    nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, stride=1),
    nn.BatchNorm2d(num_features=16),
    nn.MaxPool2d(kernel_size=2, stride = 2),
    nn.Flatten(),
    nn.BatchNorm1d(num_features=16*13*13),
    nn.Linear(in_features=16*13*13, out_features=256),
    nn.BatchNorm1d(num_features=256)
)
```

```python
summary(model, input_size=(1,28,28), batch_size=2, device='cpu')
```

# Drop out

```python
model = nn.Sequential(
    nn.Linear(in_features=4, out_features=3),
    nn.Dropout(p=0.7),
    nn.Linear(in_features=3, out_features=2)
)
```

```python
X = torch.rand(3,4)
X.shape
```

```python
model[0].weight
```

```python
criterion = nn.CrossEntropyLoss()
```

```python
optimizer = torch.optim.Adam(model.parameters())
```

```python
logits = model(X)
logits
```

```python
y = torch.tensor([1,0,1],dtype = torch.long)
```

```python
loss = criterion(logits,y)
```

```python
loss.backward()
```

```python
optimizer.step()
```
